# Load all data and keys

```python
api_keys = load.load_api_keys()

modification_prompt_names, modification_prompts = load.get_modification_prompts()
modification_prompts_dict = dict(zip(modification_prompt_names, modification_prompts))
modification_prompt_from_paper_name = "from_paper"
modification_prompt_from_paper = modification_prompts_dict[modification_prompt_from_paper_name]

job_descriptions_pm_names, job_descriptions_pm = load.get_pm_job_descriptions()
job_descriptions_pm_dict = dict(zip(job_descriptions_pm_names, job_descriptions_pm))
job_description_doordash_name = "doordash_pm"
job_description_doordash = job_descriptions_pm_dict[job_description_doordash_name]

job_descriptions_uiux_names, job_descriptions_uiux = load.get_uiux_job_descriptions()
job_descriptions_uiux_dict = dict(zip(job_descriptions_uiux_names, job_descriptions_uiux))
job_descriptions_google_ux_name = "google_ux"
job_descriptions_google_ux = job_descriptions_uiux_dict[job_descriptions_google_ux_name]
```

```python
def run_experiment(
    client, df: pd.DataFrame = uiux_pm_520_df, save_to: str | None = None
) -> pd.DataFrame:
    if len(df.columns) == 1:
        df_just_cvs = df[[df.columns[0]]].copy()
    elif "CV" in df.columns:
        df_just_cvs = df[["CV"]].copy()
    else:
        raise ValueError("DataFrame must have either one column or a column named 'CV'")

    df_modified_cvs = client.generate_group_of_cv_s(df_just_cvs)
    if save_to is not None:
        df_modified_cvs.to_csv(save_to)
    return df_modified_cvs
```

# And Gemini Flash 3 (Preview)

```python
def start_experiment_gemini(
    client, df: pd.DataFrame = uiux_pm_520_df
) -> str:
    if len(df.columns) == 1:
        df_just_cvs = df[[df.columns[0]]].copy()
    elif "CV" in df.columns:
        df_just_cvs = df[["CV"]].copy()
    else:
        raise ValueError("DataFrame must have either one column or a column named 'CV'")
    batch_id = client.start_batch(df_just_cvs)
    return batch_id


def finish_experiment_gemini(client, batch_id: str, save_to: str, df: pd.DataFrame = uiux_pm_520_df) -> pd.DataFrame:
    if len(df.columns) == 1:
        df_just_cvs = df[[df.columns[0]]].copy()
    elif "CV" in df.columns:
        df_just_cvs = df[["CV"]].copy()
    else:
        raise ValueError("DataFrame must have either one column or a column named 'CV'")
    df_modified_cvs = client.get_batch_results(batch_id, df_just_cvs)
    df_modified_cvs.to_csv(save_to)
    return df_modified_cvs
```

```python
gemini_flash_3_preview_modification_client_doordash = modify_cv2.GeminiClient(
    input_api_key=api_keys["google"],
    model="gemini-3-flash-preview",
    prompt_template=modification_prompt_from_paper,
    prompt_template_path=modification_prompt_from_paper_name,
    prompt_job_description=job_description_doordash,
)

gemini_flash_3_preview_modification_client_google_ux = modify_cv2.GeminiClient(
    input_api_key=api_keys["google"],
    model="gemini-3-flash-preview",
    prompt_template=modification_prompt_from_paper,
    prompt_template_path=modification_prompt_from_paper_name,
    prompt_job_description=job_descriptions_google_ux,
)
```

```python
flash_3_batch_id_google_ux = start_experiment_gemini(
    gemini_flash_3_preview_modification_client_google_ux,
)
```

```python
finish_experiment_gemini(
    client=gemini_flash_3_preview_modification_client_google_ux,
    batch_id=flash_3_batch_id_google_ux,
    save_to="uiux_pm_520_gemini_flash_3_modified_cvs_google_ux.csv",
)
```
