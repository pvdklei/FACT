```python
import datasets
import pandas as pd
```

# First we load the datasets from Hugging Face

```python
candidateprofiles = datasets.load_dataset("lang-uk/recruitment-dataset-candidate-profiles-english")
jobdescriptions = datasets.load_dataset("lang-uk/recruitment-dataset-job-descriptions-english")
```

# Make Pandas DataFrames from the datasets

```python
jobdescriptions_df = jobdescriptions["train"].to_pandas()
candidateprofiles_df = candidateprofiles["train"].to_pandas()
```

# We take all UI/IX and Project Manager CVs

```python

uiux_df = candidateprofiles_df[candidateprofiles_df["Position"].str.contains("UI/UX Designer", na=False)]
pm_df = candidateprofiles_df[candidateprofiles_df["Position"].str.contains("Project Manager", na=False)]
```

# Take the first 260 from both

```python
uiux_first_260_df = uiux_df.head(260).copy()
pm_first_260_df = pm_df.head(260).copy()
```

# And concatenate them

```python
uiux_first_260_df.loc[:, "Position Group"] = "UI/UX Designer"
pm_first_260_df.loc[:, "Position Group"] = "Project Manager"

uiux_pm_520_df = pd.concat([uiux_first_260_df, pm_first_260_df])
```

# Save to a CSV file

```python
uiux_pm_520_df.to_csv("uiux_pm_520_original_cvs.csv")
```
