```python
import pandas as pd
from pathlib import Path
from functools import reduce

# Folder containing the per-LLM score CSVs
scores_dir = Path("/Users/nicky/Desktop/FACT/Nicky/Ukrainian/ScoredBenteUK")

files = scores_dir.glob("*.csv")

dfs = []

for fp in files:
    fp = Path(fp)
    model_name = fp.stem # filename without ".csv""
    df = pd.read_csv(fp)

    # first column = id, second column = score
    id_col = df.columns[0]
    score_col = df.columns[1]

    df = df[[id_col, score_col]].rename(
        columns={
            id_col: "id",
            score_col: f"{model_name} resumes"
        }
    )

    dfs.append(df)

# merge all on id
merged = reduce(lambda left, right: pd.merge(left, right, on="id", how="outer"), dfs)

# optional: sort by id
merged = merged.sort_values("id")

merged.to_csv("UKScoresGG.csv", index=False)
print("Saved: Scores-newGG_MERGED.csv")
```

```python
import pandas as pd

src = pd.read_csv("/Users/nicky/Desktop/FACT/Nicky/Scored_520/Combined_UI/ScoresUIUX_Original_File_merged.csv")                 # testUX
dst = pd.read_csv("DatasetsFinal/NOrderUX.csv")      # NScores

id_col = "id"
score_col = src.columns[1]  # 3rd column in testUX (index 2)

# clean IDs
src[id_col] = src[id_col].astype("string").str.strip()
dst[id_col] = dst[id_col].astype("string").str.strip()

# choose a name for the added column (avoid name collisions)
new_col = "CV_doordash_pm_Score"

# if it already exists in dst, drop it to avoid _x/_y columns
dst = dst.drop(columns=[new_col], errors="ignore")

# merge by id
dst = dst.merge(
    src[[id_col, score_col]].rename(columns={score_col: new_col}),
    on=id_col,
    how="left"
)

# IDs not found in testUX -> 0
dst[new_col] = pd.to_numeric(dst[new_col], errors="coerce").fillna(0)

dst.to_csv("DatasetsFinal/NOrderUX.csv", index=False)
```
